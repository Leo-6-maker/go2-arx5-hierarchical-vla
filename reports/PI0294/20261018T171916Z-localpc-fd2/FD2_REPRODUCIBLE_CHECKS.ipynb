```python
import json
from pathlib import Path
p=Path('.')
m=json.loads((p/'FD2_ALL_EIGHT_ACTUAL_RESULTS.json').read_text())
r=json.loads((p/'FD2_RESOURCE_FINAL.json').read_text())
pairs=json.loads((p/'FD2_FOUR_PAIRED_INITIAL_IDENTITIES.json').read_text())['pairs']
g=json.loads((p/'FD2_SAVED_GEOMETRY_QUALIFICATION.json').read_text())
assert len(m['rows'])==m['actual_cells']==m['planned_cells']==8
assert len({x['id'] for x in m['rows']})==8
assert len(pairs)==4 and all(x['initial_native_and_scene_XML_equal'] and x['initial_two_camera_pixels_equal'] for x in pairs)
for margin in [0.004,0.012]:
 rows=[x for x in m['rows'] if x['nominal_clearance_m']==margin]
 assert len(rows)==4
 assert sum(x['original_score']['persistent_success'] for x in rows)==m['per_clearance'][str(margin)]['task_successes']
 assert all(x['audit_status']=='PASS_INDEPENDENT_SAVED_BYTE_AUDIT' and all(x['audit_checks'].values()) for x in rows)
assert len(g['rows'])==8 and g['integrations']==g['Teacher_episodes']==0
assert max(x['maximum_vertical_gap_reconstruction_error_m'] for x in g['rows'])<=1e-10
assert r['physics_ticks']==r['live_ticks']+r['shadow_ticks']<=300000
assert r['engine_starts']<=700 and r['actual_Core_accounted_CPU_s']<=28800
assert r['local_CPU_conservative_upper_s']<=7200 and r['network_accounted_conservative_upper_bytes']<=2147483648
assert r['lease']=='CLOSED' and r['future_reservations']==r['unknown_effects']==0
assert r['new_GPU_PBS_neural_training']==0
print('PASS: eight actual cells, four pairs, preserved gates and closed finite resources')
```
